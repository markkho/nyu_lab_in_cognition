Be aware that there are pros and cons to within-subject designs. One pro is that they tend to be statistically more powerful. One con is that they usually introduce potential "order effects." For example, if our participants are taking the cognitive test twice (once on, once off Ritalin) could it be that they'll do better the second time they take it (independently of whether they are on Ritalin or not)? Thus, it would be typical introduce a "counterbalancing" factor that controlled for that: Have half the participants take Ritalin first then the placebo and then reverse that order for the other half. The hope would be that any effect of taking the cognitive test first or last would then get averaged out over subjects. But this is a research design question best left for another time.
